# Base de datos SQLite paso a paso

Este notebook crea la base de datos a partir de los archivos `.sql` 19 tablas 

## Paso 1: Importar librerías y configurar rutas

```python
import sqlite3
import glob
import os

SQL_DIR = "migrations"            
DB_PATH = "database.db"    

print("Carpeta SQL existe:", os.path.isdir(SQL_DIR))
```

## Paso 2: Listar los archivos `.sql`

```python
archivos = sorted(glob.glob(os.path.join(SQL_DIR, "[0-2][0-9]_*.sql")))
for a in archivos:
    print(os.path.basename(a))
print(f"\nTotal: {len(archivos)} archivos")
assert archivos, "No se encontraron .sql. Revisa SQL_DIR."
```

## Paso 3: Crear la base de datos
Borra `database.db` si ya existía, para empezar limpio.

```python
if os.path.exists(DB_PATH):
    os.remove(DB_PATH)

con = sqlite3.connect(DB_PATH)
con.execute("PRAGMA foreign_keys = ON")

for ruta in archivos:
    with open(ruta, encoding="utf-8") as f:
        con.executescript(f.read())
    print("OK ", os.path.basename(ruta))

con.commit()
```

## Paso 4: Verificar las tablas creadas

```python
tablas = [r[0] for r in con.execute(
    "SELECT name FROM sqlite_master WHERE type='table' AND name NOT LIKE 'sqlite_%' ORDER BY name")]
print(len(tablas), "tablas:")
for t in tablas:
    print(" -", t)
```

## Paso 5: Cerrar la conexión

```python
con.close()
print("Listo. Tu base está en:", os.path.abspath(DB_PATH))
```
